```python
#5. Create a 2D Array:
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print(temperatures)
```

```python
#6. Inspect and Slice the 2D Array:
#  shape
print("Shape of the array:", temperatures.shape)

#  data type
print("Data type of the array:", temperatures.dtype)

#  total number of elements
print("Total number of elements:", temperatures.size)
```

```python
# Extract Week 1 temperatures
print("Week 1 Temperatures:", temperatures[0])

# Extract Week 2 temperatures
print("Week 2 Temperatures:", temperatures[1])
```

```python
# Extract weekend temperatures for both weeks
print("Weekend Temperatures for Week 1:", temperatures[0, -2:])
print("Weekend Temperatures for Week 2:", temperatures[1, -2:])
```

```python
import pandas as pd

#1. Create a Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)


print("Student Marks:")
print(marks)
```

```python
# Access the 1st rank student's mark
print("1st Rank Mark:", marks.iloc[0])
```

```python
# Retrieve top 3 ranks using loc
print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

```python
# Retrieve the 3rd rank student's mark
print("3rd Rank Mark:", marks.iloc[2])
```

```python
# Filter marks greater than 90
print("Ranks with marks greater than 90:")
print(marks[marks > 90])
```

```python
# Modify the 1st rank student's mark
marks.loc['Rank1'] = 100

print("Updated Marks:")
print(marks)
```

```python
# Remove the last rank entry
marks = marks.drop('Rank5')

print("Marks after removing the last rank:")
print(marks)
```

```python
# Calculate CGPA
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

```python
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

# Create Pandas DataFrame
transactions = pd.DataFrame(data)

# Display the DataFrame
print(transactions)
```

```python
# Display the complete DataFrame
print(transactions)


print(transactions.head())


print(transactions.tail())

# Display the shape 
print(transactions.shape)

# Display column names
print(transactions.columns)

# Display data types
print(transactions.dtypes)

# Display basic information
print(transactions.info())
```

```python
print(transactions[['ProductCategory', 'Amount']])
```

```python
# Using iloc
print(transactions.iloc[:, -3:])

```

```python
# Using loc
print(transactions.loc[:, ['ProductCategory', 'Region', 'Amount']])
```

```python
result = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(result)
```

```python
print(transactions['ProductCategory'].value_counts())
```

```python
print(transactions['Region'].unique())
```

```python
result = transactions.groupby('Region')['Amount'].mean()

print(result)
```

```python
# Update Amount 
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

# Display updated DataFrame
print(transactions)
```

```python
# Add Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

# Display DataFrame
print(transactions)
```

```python
# Remove the row where TransactionID is 109
transactions = transactions[
    transactions['TransactionID'] != 109
]

# Display updated DataFrame
print(transactions)
```

```python
# Delete the Discount column
transactions = transactions.drop('Discount', axis=1)

# Display the final DataFrame
print(transactions)
```
